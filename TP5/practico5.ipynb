```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(123)
legajos_generados = ['TUP-'+ str(i) for i in range(1001, 1021)]
carreras_disponibles = ['Programacion','Redes','Sistemas']

datos_academicos = {
    'Legajo': legajos_generados,
    'Carrera': np.random.choice(carreras_disponibles, 20),
    'Matematica': np.random.randint(2, 11, 20),
    'Bases_Datos': np.random.randint(2, 11, 20),
    'Algoritmos': np.random.randint(2, 11, 20)
}

df_generador = pd.DataFrame(datos_academicos)
df_generador.to_csv('notas_semestre.csv', index=False)
print('Archivo notas_semestre.csv generado con éxito.')

# Leer el archivo CSV en una nueva celda
df_notas = pd.read_csv('notas_semestre.csv')

# Establecer 'Legajo' como índice
df_notas = df_notas.set_index('Legajo')

# Mostrar las primeras 5 filas para verificar
print("Primeras 5 filas con Legajo como índice:")
display(df_notas.head())
```

```python
# Auditoria estadistica de las notas
print("Resumes estadístico de las materias:")
display(df_notas.describe())

# Extraer decima fila
fila_10 = df_notas.iloc[9]
print("\n Decima fila: ")
display(fila_10)

#extraer primeras 3 filas
subset = df_notas.iloc[0:3, 0:2]
print("Primeras 3 filas y primeras 2 columnas")
display(subset)
```

```python
# Modificar nota 
df_notas.loc['TUP-1005', 'Bases_Datos'] = 9
print("Nota de base de datos actualizada a 9")

df_notas.loc['TUP-1018', 'Matematica'] = 7
print("Nota de matematicas actualizada a 7")

# Verificar cambios 
verificacion = df_notas.loc[['TUP-1005', 'TUP-1018']]
print("Verificacion de cambio:")
display(verificacion)
```

```python
# Mascara 
mascara = (df_notas['Carrera'] == 'Programacion') & (df_notas['Algoritmos'] >= 8)

#aplicar mascara 
destacados = df_notas[mascara]

# Mostrar datos 
print("Alumnos destacados: ")
display(destacados)
```

```python
# Filtrar alumnos de redes
redes = df_notas[df_notas['Carrera'] == 'Redes']

# Extraer columna de matematicas
notas_matematica = redes['Matematica']

# Calcular el promedio
promedio = notas_matematica.mean()
print(f"Promedio de notas de Matematica para alumnos de Redes: {promedio:.2f}")
```
