```python
import pandas as pd
import csv

datos_tienda = [ 
    {"ID": 1001, "Producto": "Laptop Gamer", "Categoria": "Computacion", "Precio": 1200000, "Stock": 15}, 
    {"ID": 1002, "Producto": "Mouse Inalambrico", "Categoria": "Accesorios", "Precio": 25000, "Stock": 100}, 
    {"ID": 1003, "Producto": "Teclado Mecanico", "Categoria": "Accesorios", "Precio": 85000, "Stock": 40}, 
    {"ID": 1004, "Producto": "Monitor 27 Pulgadas", "Categoria": "Computacion", "Precio": 350000, "Stock": 25},
    {"ID": 1005, "Producto": "Auriculares Bluetooth", "Categoria": "Audio", "Precio": 60000, "Stock": ""},
    {"ID": 1006, "Producto": "Silla Ergonomica", "Categoria": "Muebles", "Precio": 250000, "Stock": 10} 
] 
with open('productos.csv', 'w', newline='', encoding='utf-8') as archivo: 
    columnas = ["ID", "Producto", "Categoria", "Precio", "Stock"] 
    writer = csv.DictWriter(archivo, fieldnames=columnas) 
    writer.writeheader() 
    writer.writerows(datos_tienda) 
print("Archivo 'productos.csv' creado exitosamente.") 
```

```python
df_productos = pd.read_csv ('productos.csv')

#mostrar las primeras 3 filas 
print ("Primeras 3 filas: ")
display (df_productos.head(3))

#mostrar las ultimas 2 filas
print ("\n Ultimas 2 filas: ")
display (df_productos.tail(2))
```

```python
#auditoria de calidad 
print ("Auditoria de calidad: ")
display (df_productos.info())

#contar valores no nulos en stock
valores_no_nulos = df_productos['Stock'].count()
total_filas = len(df_productos)
faltantes = total_filas - valores_no_nulos

print (f"\n Columna stock: {valores_no_nulos} valores no nulos de {total_filas} totales")
print (f"Se detectaron {faltantes} datos faltantes")

#auditoria financiera
print ("\n Auditorias financiera: ")
display (df_productos.describe().round(2))

precio_promedio = df_productos['Precio'].mean()
prducto_mas_barato = df_productos.loc[df_productos['Precio'].idxmin(), 'Producto']
precio_minimo = df_productos['Precio'].min()

print (f"\n Precio promedio: {precio_promedio:,.2f}")
print (f"Producto mas barato: '{prducto_mas_barato}' (${precio_minimo:,.2f})")
```

```python
#extraer producto y precio
tabla_marketing = df_productos[["Producto", "Precio"]]
print ("DataFrame para marketing: ")
display (tabla_marketing)
print (f"Tipo de dato: {type(tabla_marketing)}")

#extraer solo precio como serie 
serie_precios = df_productos["Precio"]
print ("Sere de precios para logistica: ")
display (serie_precios)
print (f"Tipo de dato: {type(serie_precios)}")
```

```python
#capital por produto
capital_producto = df_productos["Precio"] * df_productos["Stock"]
print("Capital potencial por producto")
display(capital_producto)
```
