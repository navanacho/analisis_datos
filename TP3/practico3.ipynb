```python
#ejercicio 2
ventas = np.array([
    [1500, 2000, 1800, 2200],
    [1100, 1300, 1200, 1500],
    [3000, 3100, 2900, 3500]
])

#centa total
ventaTotal = np.sum(ventas)
print(ventaTotal)

#promedio por semana
promedioSemanal = np.mean(ventas, axis = 0)
print(promedioSemanal)

#venta max por sucursal
maxSucursal = np.max(ventas, axis = 1)
print(maxSucursal)
```

```python
#ejercicio 3
numeros = np.arange(1,21)
print (numeros)

#mascara de pares
mascaraPares = numeros % 2 == 0
print(mascaraPares)

#aplicar mascara
pares = numeros[mascaraPares]
print(pares)
```

```python
#ejercicio 4
edades = np.array([22, 55, 34, 19, 45, 61, 28, 30, 42])

#personas netre 30 y 50
filtro = (edades >=30) & (edades <= 50)
resultado = edades[filtro]
print(resultado)

#menores de 25 o mayores de 55
segundoFiltro = (edades < 25) | (edades > 55)
segundoResultado = edades[segundoFiltro]
print(segundoResultado)
```
