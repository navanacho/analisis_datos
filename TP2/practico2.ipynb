```python
import numpy as np
```

```python
edades = np.array([22, 25, 30, 35, 40])

print("Dimensiones:", edades.ndim)
print("Forma:", edades.shape)
print("Tipo de datos:", edades.dtype)

edadesFloat = edades.astype(np.float64)
print("Array con tipo de datos float64:", edadesFloat)
print("Nuevo tipo de dato:", edadesFloat.dtype)
```

```python
#Ejercicio 2
idClientes = np.arange(1, 101)

puntosGraficos = np.linspace(0, 10, 50)

print("Primeros 10IDs:", idClientes[:10])
print("Primeros 10 puntos gráficos:", puntosGraficos[:10])
```

```python
#Ejercicio 3
matrizAleatoria = np.random.rand(3, 4)

matrizDeUnos = np.ones((3,4))

print("Matriz aleatoria:\n", matrizAleatoria)
print("Matriz de unos:\n", matrizDeUnos)
```

```python
#Ejercicio 4
sucursales = int(input("Ingrese el número de sucursales: "))
productos = int(input("Ingrese el número de productos: "))

inventario = np.zeros((sucursales, productos), dtype = int)

print ("\nBase de datos de inventario inicializada correctamente")
print("Matriz de inventario (sucursales x productos):\n", inventario)
```

```python
#Ejercicio 5 
vector = np.arange(1, 13)
print ("Vector original:", vector)

matriz = vector.reshape(3, 4)
print("Matriz reshaped (3x4):\n", matriz)
print ("Nueva matriz despues del reshape:", matriz.ndim)
```

```python
#Ejercicio 6
temperaturas = np.array([15.5, 20.3, 25.1, 30.0, 35.2])
print("Tipo de datos original:", temperaturas.dtype)

temperaturasInt = temperaturas.astype(np.int32)
print("Array con tipo de datos int32:", temperaturasInt)
print("Nuevo tipo de dato:", temperaturasInt.dtype)
```

```python
#Ejercicio 7
tiradas = np.random.randint(low=1, high=7, size=10)
print("Tiradas de dados:", tiradas)
```

```python
#Ejercicio 8 
humedadTexto = np.array(["45.2", "48.1", "42.5", "50.0"])

humedadNumerica = humedadTexto.astype(np.float64)
print("Suma de humedad (como números):", humedadNumerica.sum())
```

```python
#Ejercicio 9 
matriz = np.eye(4)
print("Matriz identidad 4x4:\n", matriz)

resultado = matriz * 5
print("Matriz identidad multiplicada por 5:\n", resultado)
```
