```python
producto = {
    "id": 1000,
    "nombre": "monitor 24 pulgadas",
    "precio": 152222,
    "stock": 15
}

# 1 Restar 1 al stock
producto["stock"] = producto["stock"] - 1

# 2 Agregar categoria
producto["categoria"] = "Tecnología"

# 3 Mensaje dinámico
print("Se vendió un", producto["nombre"], ". Quedan", producto["stock"], "en stock.")
```

```python
ventas = [
    {"ticket": 1, "monto": 5000, "metodo": "Efectivo"},
    {"ticket": 2, "monto": 12000, "metodo": "Tarjeta"},
    {"ticket": 3, "monto": 8500, "metodo": "Efectivo"},
    {"ticket": 4, "monto": 3000, "metodo": "MercadoPago"}
]

totalDia = 0
totalEfectivo = 0

for venta in ventas:
    totalDia += venta["monto"]
    
    if venta["metodo"] == "Efectivo":
        totalEfectivo += venta["monto"]

print("Total recaudado en el día:", totalDia)
print("Total recaudado en efectivo:", totalEfectivo)
```
