```python
import pandas as pd

df = pd.read_csv('streams_sucios.csv')

print("Archivo cargado exitosamente")
print(f"Dimensiones iniciales: {df.shape[0]} filas × {df.shape[1]} columnas")
display(df.head())
```

```python
# Contar valores nulos por columna
print("📋 Valores nulos por columna:")
display(df.isnull().sum())

# Contar filas duplicadas exactas
duplicados = df.duplicated().sum()
print(f"\n🔄 Filas duplicadas exactas: {duplicados}")

# Total de filas antes de limpiar
print(f"\n📈 Total de filas iniciales: {len(df)}")
```

```python
# Eliminar filas donde la columna 'Canal' es nula
df = df.dropna(subset=['Canal'])

print(f"Filas sin 'Canal' eliminadas. Filas restantes: {len(df)}")
```

```python
# Rellenar nulos en 'Categoria' con texto por defecto
df['Categoria'] = df['Categoria'].fillna('Sin Clasificar')

print("Columna 'Categoria' completada con 'Sin Clasificar'")
```

```python
# Rellenar nulos en 'Espectadores_Pico' con 0 (valor numérico)
df['Espectadores_Pico'] = df['Espectadores_Pico'].fillna(0)

print("Columna 'Espectadores_Pico' completada con 0")
```

```python
# Verificar que no queden nulos
print("Valores nulos restantes:")
print(df.isnull().sum())

# Verificar tamaño final
print(f"\n Tamaño final del DataFrame: {df.shape[0]} filas × {df.shape[1]} columnas")

# Mostrar resumen rápido
print("\n Primeras 5 filas del dataset limpio:")
display(df.head())
```

```python
# 1. Filtrar solo categoría 'Gaming'
df_gaming = df[df['Categoria'] == 'Gaming']

# 2. Encontrar el récord máximo de espectadores en Gaming
max_espectadores = df_gaming['Espectadores_Pico'].max()

# 3. Identificar al streamer que logró ese récord
streamer_record = df_gaming.loc[df_gaming['Espectadores_Pico'].idxmax(), 'Canal']

# 4. Armar el informe gerencial con print() redactado
print("REPORTE GERENCIAL - RECOMENDACIÓN DE PATROCINIO")
print("=" * 60)
print(f"Categoría analizada: Gaming")
print(f"Streamers en esta categoría: {len(df_gaming)}")
print(f"Récord de espectadores: {max_espectadores:,.0f}")
print(f"Streamer destacado: {streamer_record}")
print("-" * 60)
print(f"RECOMENDACIÓN: Se sugiere priorizar el patrocinio oficial")
print(f"   de '{streamer_record}', quien alcanzó el máximo histórico")
print(f"   de audiencia en la categoría Gaming con {max_espectadores:,.0f} espectadores.")
print("=" * 60)
```
