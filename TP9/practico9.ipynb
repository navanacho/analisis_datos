```python
import pandas as pd 
import numpy as np 

meses = ['Ene','Feb','Mar','Abr','May','Jun','Jul','Ago','Sep','Oct','Nov','Dic']
np.random.seed(42)

datos_it ={
    'Mes': meses,
    'Sueldo_Promedio_USD': [1100, 1150, 1200, 1180, 1250, 1300, 1350, 1400, 1380, 1450, 1500, 1550],
    'Puestos_Abiertos': np.random.randint(50, 200, 12)
}

df_it = pd.DataFrame(datos_it)
df_it.to_csv('mercado_it_2025.csv', index=False)
print("Archivo mercado_it_2025.csv generado con exito")
```

```python
#Ejercicio 2
import matplotlib.pyplot as plt 

df_it = pd.read_csv('mercado_it_2025.csv')

plt.plot(df_it['Mes'], df_it['Sueldo_Promedio_USD'], color='green', marker='o')

plt.title('Evolución del Sueldo Promedio en USD (2025)')
plt.xlabel('Mes')
plt.ylabel('Sueldo Promedio (USD)')
plt.grid(True)

plt.show()
```

```python
#Ejercicio 3
plt.bar(df_it['Mes'], df_it['Puestos_Abiertos'], color='orange')

plt.title('Cantidad de Puestos Abiertos por Mes')
plt.xlabel('Mes')
plt.ylabel('Puestos Abiertos')
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()

mes_max_demanda = df_it.loc[df_it['Puestos_Abiertos'].idxmax(), 'Mes']
print(f"El mes con mayor demanda de puestos fue: {mes_max_demanda}")
```

```python
#Ejercicio 4
plt.style.use('seaborn-v0_8')

plt.figure(figsize=(12 ,6))

plt.plot(df_it['Mes'], df_it['Sueldo_Promedio_USD'], color='green', marker='o', linewidth=2)
plt.title('Evolución del Sueldo Promedio en USD (2025)')
plt.xlabel('Mes')
plt.ylabel('Sueldo Promedio (USD)')

promedio_anual = df_it['Sueldo_Promedio_USD'].mean()
plt.axhline(y=promedio_anual, color='red', linestyle='--', label=f'Promedio: ${promedio_anual:.2f}')
plt.legend()

plt.grid(True)
plt.show()
```


INFORME DE GESTIÓN:
Los gráficos muestran una tendencia clara y sostenida de crecimiento en el sueldo promedio a lo largo del año, 
pasando de 1.100 USD a 1.550 USD. Por otro lado, la cantidad de puestos abiertos presenta fluctuaciones 
mensuales sin un patrón ascendente constante. Esto indica que no existe una correlación directa entre 
aumento de vacantes y subida salarial. El crecimiento de los sueldos responde principalmente a la 
escasez de perfiles técnicos calificados y a la competencia entre empresas por retener talento, 
mientras que la contratación sigue ciclos presupuestarios o de proyectos puntuales. En conclusión, 
el mercado IT local está en fase de maduración donde el valor del profesional especializado crece 
incluso con una demanda de puestos estable o variable.
